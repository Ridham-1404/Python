```python
#Q1
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original temperatures:", temperature)

corrected_temperature = temperature + 2
print("Corrected temperatures (+2°C):", corrected_temperature)

fahrenheit = (temperature * 9 / 5) + 32
print("Temperatures in Fahrenheit:", fahrenheit)

above_32 = temperature[temperature > 32]
print("Readings greater than 32°C:", above_32)

count_above_32 = np.sum(temperature > 32)
print("Count of readings exceeding 32°C:", count_above_32)
```

```python
#Q2
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
print("Steps matrix:\n", steps)

total_steps = np.sum(steps)
print("Total steps recorded:", total_steps)

mean_steps = np.mean(steps)
print("Mean number of steps:", mean_steps)

max_steps = np.max(steps)
min_steps = np.min(steps)
print("Maximum steps:", max_steps)
print("Minimum steps:", min_steps)

total_per_day = np.sum(steps, axis=0)
print("Total steps per day:", total_per_day)

total_per_user = np.sum(steps, axis=1)
print("Total steps per user:", total_per_user)

max_position = np.unravel_index(np.argmax(steps), steps.shape)
user_index, day_index = int(max_position[0]), int(max_position[1])
print("Position (user, day) with maximum steps:", (user_index, day_index))
```

```python
#Q3
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
print("Original array:", original)

subset = original[1:5]
print("Subset:", subset)

subset[0] = 999
print("Original after modifying subset:", original)
print("Subset after modification:", subset)

subset_copy = original[1:5].copy()
subset_copy[0] = 500
print("Original after modifying copied slice:", original)
print("Copied slice after modification:", subset_copy)

matrix = np.arange(1, 13).reshape(3, 4)
print("3x4 matrix:\n", matrix)

first_row = matrix[0, :]
last_row = matrix[-1, :]
second_column = matrix[:, 1]
sub_matrix = matrix[1:3, 2:4]

print("First row:", first_row)
print("Last row:", last_row)
print("Second column:", second_column)
print("Rows 1-2, columns 2-3:\n", sub_matrix)

flattened = matrix.flatten()
raveled = matrix.ravel()
print("Flattened array:", flattened)
print("Raveled array:", raveled)

raveled[0] = 1000
print("Matrix after modifying raveled array:\n", matrix)
print("Raveled array is a view, original changed:", matrix.flatten()[0] == 1000)

flattened[0] = 2000
print("Matrix after modifying flattened array:\n", matrix)
print("Flattened array is a copy, original unchanged:", matrix[0, 0] != 2000)

print("Shape:", matrix.shape)
print("Number of dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Dtype:", matrix.dtype)
```

```python
#Q4
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

X_T = X.T
print("Transpose of X:\n", X_T)

XtX = X.T @ X
print("X.T @ X:\n", XtX)

XtX_inv = np.linalg.inv(XtX)
print("Inverse of X.T @ X:\n", XtX_inv)

beta = XtX_inv @ X.T @ y
print("OLS coefficients (beta):", beta)

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score for new user:", predicted_score)
```
