```python
#Loops
#1 while
i = 1
while i<=10 :
      print(i)
      i=i+1

#Range Function
print("range(10)           --> ", list(range(10)))
print("range(10, 20)       --> ", list(range(10, 20)))
print("range(-20, -10, 2)  --> ", list(range(-20, -10, 2)))
print("range(0, 2, 20)     --> ", list(range(0, 2, 20)))
print("range(-20, -10, -2) --> ", list(range(-20, -10, -2)))
```

```python
#2 for loop - version 1

for i in range(0, 10):
          print(i)


#2 for loop - version 2

for i in range(0, 2, 20):
          print(i)

#3 for loop - version 3

for i in range(0, -10, -1):
          print(i)

```

```python
#Print table for 5
for i in range(1,21):
          print(5, "*", i, "=", i*5 )



#Print the sum of all numbers from 1 to 10
#M1
s = 0
for i in range(1, 11):
          s=s+i
          print("Sum = ", s)

#M2
print("Sum is ", sum(range(1,11)))


```

```python
#Table of n
n = input("Enter the number : ")
for i in range(1, 21):
        print(n, "*", i, "=", int(n)*i)

#Sum of n numbers
n=int(input("Enter the number : "))
s = 0
for i in range(1,n):
          s = s+i
          print("sum = ", s)
```
