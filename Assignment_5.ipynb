```python
#Q1
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print("Original array : ", arr)
print("Addition of 2 : ", arr+2)
print("Multiplication by 3 : ", arr * 3)
print("Division by 2 : ", arr / 2)
```

```python
#Q2
# Q2(a)
arr2 = np.array([1, 2, 3, 6, 4, 5])
print("Reversed:", arr2[::-1])

# Q2(b)
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

for name, array in [("x", x), ("y", y)]:
    values, counts = np.unique(array, return_counts=True)
    most_frequent = values[np.argmax(counts)]
    indices = np.where(array == most_frequent)[0]
    print(f"{name}: value = {most_frequent}, indices = {indices}")

# Q3
arr3 = np.array([[10, 20, 30],
                 [40, 50, 60],
                 [70, 80, 90]])

print("1st row, 2nd column:", arr3[0, 1])
print("3rd row, 1st column:", arr3[2, 0])


# Q4
ridham = np.linspace(10, 100, 25)

print("Dimensions:", ridham.ndim)
print("Shape:", ridham.shape)
print("Total elements:", ridham.size)
print("Data type:", ridham.dtype)
print("Bytes consumed:", ridham.nbytes)

transpose_ridham = ridham.reshape(-1, 1)
print("Transpose using reshape:\n", transpose_ridham)

print("Using T attribute:", ridham.T)


# Q5
ucs420_ridham = np.array(
    [10, 20, 30, 40,
     50, 60, 70, 80,
     90, 15, 20, 35]
).reshape(3, 4)

print("Mean:", np.mean(ucs420_ridham))
print("Median:", np.median(ucs420_ridham))
print("Maximum:", np.max(ucs420_ridham))
print("Minimum:", np.min(ucs420_ridham))
print("Unique elements:", np.unique(ucs420_ridham))

reshaped_ucs420_ridham = ucs420_ridham.reshape(4, 3)
print("Reshaped array:\n", reshaped_ucs420_ridham)

resized_ucs420_ridham = np.resize(ucs420_ridham, (2, 3))
print("Resized array:\n", resized_ucs420_ridham)
```
