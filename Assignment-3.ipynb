## Q4. Read a CSV file and display its first five rows

```python
import os

iris_file = "Iris.csv"

# If the Kaggle Iris.csv is not available, create a local Iris CSV
# so that the remaining code runs without an internet connection.
if not os.path.exists(iris_file):
    try:
        from sklearn.datasets import load_iris

        iris = load_iris()
        iris_df = pd.DataFrame(
            iris.data,
            columns=["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
        )
        iris_df["Species"] = [iris.target_names[i] for i in iris.target]
        iris_df.insert(0, "Id", range(1, len(iris_df) + 1))
        iris_df.to_csv(iris_file, index=False)
    except ImportError:
        raise FileNotFoundError(
            "Iris.csv was not found. Download the Kaggle Iris dataset and place Iris.csv "
            "in the same folder as this notebook."
        )

iris_df = pd.read_csv(iris_file)
iris_df.head()
```

## Q5. Delete row 4 and column 3 from the CSV data

```python
# Delete row at index 4 and column at positional index 3
iris_modified = iris_df.drop(index=iris_df.index[4])
iris_modified = iris_modified.drop(columns=iris_modified.columns[3])

iris_modified
```

## Q6. Create `employees.csv` and perform the required operations

```python
employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employee_data)
employees["Joining_Date"] = pd.to_datetime(employees["Joining_Date"])

employees.to_csv("employees.csv", index=False)
employees
```

### Q6(a). Shape of the DataFrame

```python
shape = employees.shape
print("Number of rows:", shape[0])
print("Number of columns:", shape[1])
```

### Q6(b). Data types and non-null counts

```python
summary = pd.DataFrame({
    "Data Type": employees.dtypes.astype(str),
    "Non-Null Count": employees.notna().sum()
})
summary
```

### Q6(c). Descriptive statistics

```python
employees.describe()
```

### Q6(d). First 5 rows and last 3 rows

```python
print("First 5 rows:")
display(employees.head(5))

print("Last 3 rows:")
display(employees.tail(3))
```

### Q6(e). Required statistics

```python
average_salary = employees["Salary"].mean()
total_bonus = employees["Bonus"].sum()
youngest_age = employees["Age"].min()
highest_rating = employees["Rating"].max()

print("Average salary:", average_salary)
print("Total bonus paid:", total_bonus)
print("Youngest employee's age:", youngest_age)
print("Highest performance rating:", highest_rating)
```

### Q6(f). Sort by Salary in descending order

```python
sorted_employees = employees.sort_values(by="Salary", ascending=False)
sorted_employees
```

### Q6(g). Add Performance_Category

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance_Category"] = employees["Rating"].apply(performance_category)
employees
```

### Q6(h). Identify missing values

```python
missing_values = employees.isnull().sum()
missing_values
```

### Q6(i). Rename `Employee_ID` to `ID`

```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)
employees
```

### Q6(j). Find employees with more than 5 years of experience and employees in IT

```python
more_than_5_years = employees[employees["Years_of_Experience"] > 5]
it_employees = employees[employees["Department"] == "IT"]

print("Employees with more than 5 years of experience:")
display(more_than_5_years)

print("Employees in the IT department:")
display(it_employees)
```

### Q6(k). Add Tax column (10% of Salary)

```python
employees["Tax"] = employees["Salary"] * 0.10
employees
```

### Q6(l). Save the modified DataFrame to a new CSV file

```python
output_file = "employees_modified.csv"
employees.to_csv(output_file, index=False)

print(f"Modified DataFrame saved as: {output_file}")
print(f"File exists: {os.path.exists(output_file)}")
```
